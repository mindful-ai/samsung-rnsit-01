### Groq Connection

```python
key_path = r"C:\mindful-ai\samsung\samsung-ai\key-vault\groq-api.key"
with open(key_path) as f:
    api_key = f.read().strip()
api_key[:5]
```

```python
from groq import Groq
```

```python
client = Groq(api_key=api_key)
```

```python
response = client.chat.completions.create(
    model='llama-3.1-8b-instant',
    messages=[
        {'role':'system', 'content':'you are a very rude professor'},
        {'role':'user', 'content': 'tell me about quantum computing'}
    ],
    temperature=0.7,
    max_tokens=500
)
```

```python
print(response.choices[0].message.content)
```

```python
response = client.chat.completions.create(
    model='llama-3.1-8b-instant',
    messages=[
        {'role':'system', 'content':'you are a kind encouraging professor'},
        {'role':'user', 'content': 'tell me about quantum computing'}
    ],
    temperature=0.7,
    max_tokens=500
)
```

```python
print(response.choices[0].message.content)
```

```python
response = client.chat.completions.create(
    model='llama-3.1-8b-instant',
    messages=[
        {'role':'system', 'content':'you are a kind encouraging professor'},
        {'role':'user', 'content': 'tell me about quantum computing as if I am a 10 year old'}
    ],
    temperature=0.7,
    max_tokens=500
)
```

```python
print(response.choices[0].message.content)
```
