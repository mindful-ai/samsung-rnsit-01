### Task: 'zero-shot-classification'

```python
model = pipeline('zero-shot-classification')
```

```python
model("This is a training program on Artificial Intelligence", candidate_labels=["sports", "education", "academics", "politics"])
```

### Task: Text Generation

```python
LM = "HuggingFaceTB/SmolLM-360M-Instruct"
```

```python
model = pipeline('text-generation', model=LM)
```

```python
text = model("gravity is a force")
```

```python
print(text[0]["generated_text"])
```
